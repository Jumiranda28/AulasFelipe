```python
class Musica:
    def __init__(self, nome, cantor, genero, duração, ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duração = duração
        self.ano = ano

        

    def mostrar_nome(self):
        print(f"Nome da música: {self.nome}")
    def mostrar_cantor(self):
            print(f"Cantor: {self.cantor}")
    def mostrar_genero(self):
            print(f"Gênero: {self.genero}")
    def mostrar_duração(self):
            print(f"Duração: {self.duração}")
    def mostrar_ano(self):
            print(f"Ano: {self.ano}")

    def mostrar_informacoes(self):
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duração()
        self.mostrar_ano()
```
