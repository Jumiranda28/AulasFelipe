# Aula pratica: Filmes

### Criando class

```python
class Filme:
    pass
```

### Criando objeto

```python
f1 = Filme()
f2 = Filme()
```

```python
f1.titulo = "Vingadores: Ultimato"
f1.duração = 181
f1.genero = "Ação / Ficção científica"

f2.titulo = "Vingadores: Guerra Infinita"
f2.duração = 149
f2.genero = "Ação / Ficção científica"
```

### Exibindo informações

```python
print(f"Nome do filme: {f1.titulo}")
print(f"Duração: {f1.duração} minutos")
print(f"Gênero: {f1.genero}")

print(f"Nome do filme: {f2.titulo}")
print(f"Duração: {f2.duração} minutos")
print(f"Gênero: {f2.genero}")
```

### Modificar as informações de um dos  filmes

```python
f2.titulo = "Vingadores: Guerra Civil"
f2.duração = 149
f2.genero = "Ação / Ficção científica / Drama"
```

### Consultando mudanças

```python
print(f"Nome do filme: {f2.titulo}")
print(f"Duração: {f2.duração} minutos")
print(f"Gênero: {f2.genero}")
```
