FUNÇÃO:

```python
def texto_colorido(texto, cor, local_cor, estilo):
    cores_fonte = {
        "preto": "\033[30m",
        "vermelho claro": "\033[91m",
        "vermelho escuro": "\033[31m",
        "verde claro": "\033[92m",
        "verde escuro": "\033[32m",
        "amarelo claro": "\033[93m",
        "amarelo escuro": "\033[33m",
        "azul claro": "\033[94m",
        "azul escuro": "\033[34m",
        "magenta claro": "\033[95m",
        "magenta escuro": "\033[35m",
        "ciano claro": "\033[96m",
        "ciano escuro": "\033[36m",
        "cinza claro": "\033[37m",
        "cinza escuro": "\033[90m",
        "branco": "\033[97m"
    }

    cores_fundo = {
        "preto": "\033[40m",
        "vermelho claro": "\033[101m",
        "vermelho escuro": "\033[41m",
        "verde claro": "\033[102m",
        "verde escuro": "\033[42m",
        "amarelo claro": "\033[103m",
        "amarelo escuro": "\033[43m",
        "azul claro": "\033[104m",
        "azul escuro": "\033[44m",
        "magenta claro": "\033[105m",
        "magenta escuro": "\033[45m",
        "ciano claro": "\033[106m",
        "ciano escuro": "\033[46m",
        "cinza claro": "\033[47m",
        "cinza escuro": "\033[100m",
        "branco": "\033[107m"
    }

    estilos = {
        "negrito": "\033[1m",
        "sublinhado": "\033[4m",
        "italico": "\033[3m",
        "nenhum": " "
    }

    if local_cor.lower() == "fonte":
        cor_codigo = cores_fonte[cor.lower()]
    elif local_cor.lower() == "fundo":
        cor_codigo = cores_fundo[cor.lower()]

    estilo_codigo = estilos[estilo.lower()]

    return f"{cor_codigo}{estilo_codigo}{texto}\033[0m"
```

Chamando função

```python
resultado = texto_colorido(texto, cor, local_cor, estilo)
print(resultado)
```

Implementando calcular mediana

Conjuntos:

```python
A = [23, 14, 31, 17, 20]
B = [75, 1, 4, 3, 19, 50, 64, 31]
```

Função

```python
def mediana(lista):
    lista_ordenada = sorted(lista)
    quant_elementos = len(lista)
    meio = quant_elementos // 2

    if quant_elementos % 2 == 0:
        return f"Quantidade de elementos par: {quant_elementos} \nMediana: {(lista_ordenada[meio - 1] + lista_ordenada[meio]) / 2}"
    else:
        return f"Quantidade de elementos impar: {quant_elementos} \nMediana: {lista_ordenada[meio]}"
```

Resultado

```python
mediana_A = mediana(A)
mediana_B = mediana(B)

print(f"Mediana da Conjunto A: {mediana_A} \nMediana da Conjunto B: {mediana_B}")
```

Moda:

Listas:


```python
Valor_A = [12, 15, 18, 15, 17]
Valor_B = [23, 14, 31, 17, 23, 20, 23, 31]
```

Função

```python
def moda(lista):
    Valor_moda = 0
    Moda = 0
    for valor in lista:
        frequencia= lista.count(valor)
        if frequencia > Valor_moda:
            Valor_moda = frequencia
            Moda = valor

    return f"A moda dessa lista foi {Moda}"
    
           
```

Chamando função

```python
Moda_A = moda(Valor_A)
Moda_B = moda(Valor_B)
print(f"{Moda_A} \n{Moda_B}")
```

Moda Caso Geral

```python
dados_1 = [10, 20, 20, 30, 40]

dados_2 = [10, 10, 20, 20, 30, 40]

dados_3 = [10, 10, 20, 20, 30, 30, 40]

dados_4 = [10, 20, 30, 40]
```

```python
def geral_moda(valores):
    modas = []
    maior_frequencia = 0
    
    for valor in valores:
        frequencia = valores.count(valor)
        
        if frequencia > maior_frequencia:
            maior_frequencia  = frequencia
            modas = [valor]
        
        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)
            
        
    if maior_frequencia == 1:
        return "Amodal", valores
    
    elif len(modas) == 1:
        return "Unimodal", modas
    
    elif len(modas) == 2:
        return "Bimodal", modas
    
    else:
        return "Multimodal", modas        
```

```python
geral_1 = geral_moda(dados_1)

geral_2 = geral_moda(dados_2)

geral_3 = geral_moda(dados_3)

geral_4 = geral_moda(dados_4)

print(geral_1)

print(geral_2)

print(geral_3)

print(geral_4)
```

variancia

```python
def calcula_variancia(lista):
    media = sum(lista)/ len(lista)
    valores = []
    for i in lista:
        sigma = (i - media) ** 2
        valores.append(sigma)
    variancia = sum(valores) / len(lista)

    return variancia      
```

```python
lista = [12, 18, 15, 21, 14, 16]

print(f"{calcula_variancia(lista):.2f}")
```

desvio padrão

```python
import math
```

```python
def desvio_padrao(lista):
    variancia = calcula_variancia(lista)
    dp = math.sqrt(variancia)

    return dp 
```

```python
print(f"{desvio_padrao(lista):.2f}")
```
