```python
tempos = [32, 48, 27, 55, 41, 29]

atrasadas = 0
soma = 0
for tempo in tempos:
    soma += 1
    if tempo <= 40:
        print(f"Entrega {soma}: O tempo de entrega é de {tempo} minutos, a entrega está dentro do prazo.")
    else:
        atrasadas += 1
        print(f"Entrega {soma}: O tempo de entrega é de {tempo} minutos, a entrega está atrasada.")

tempo_total = sum(tempos)
print(f"Tempo total de entregas: {tempo_total} minutos.")

tempo_medio = tempo_total / len(tempos)
print(f"Tempo médio de entrega: {tempo_medio:.2f} minutos.")

print(f"Total de entregas atrasadas: {atrasadas}")

percentual_atrasados = (atrasadas / soma) * 100
print(f"Percentual de entregas atrasadas: {percentual_atrasados}%")


```

```python
print(f"Controle de produção.\nPara cada turno digite a quantidade de peças produzidas. \nOs valores informados devem ser entre 300 e 800 peças.\n"
      f"Caso seja digitado um valor fora desse intervalo, o programa solicitará a digitação novamente. \n"  
      f"O programa será encerrado quando a tecla '-1' for digitada.")

producoes = []

quant_turnos = int(input("Digite a quantidade de peças produzida em cada turno:(Digite '-1' para encerrar o programa) "))

while quant_turnos != -1:
    if 300 <= quant_turnos <= 800:
        producoes.append(quant_turnos)
    else:
        print("Valor inválido. Digite um valor entre 300 e 800.")
    quant_turnos = int(input("Digite a quantidade de peças produzida em cada turno:(Digite '-1' para encerrar o programa) "))

turno_meta = 0

valor_turno = 0
for p in producoes:
    valor_turno += 1
    if p < 400:
        print(f"Turno {valor_turno}: A quantidade de peças produzidas foi de {p}. Classificação: crítico.")
    elif p > 499:
        turno_meta += 1
        print(f"Turno {valor_turno}: A quantidade de peças produzidas foi de {p}. Classificação: Meta atingida.")
    else:
        print(f"Turno {valor_turno}: A quantidade de peças produzidas foi de {p}. Classificação: Abaixo da meta.")

producao_total = sum(producoes)
print(f"Produção total: {producao_total} peças.")

print(f"Quantidade de turnos: {valor_turno}")

producao_media = producao_total / valor_turno
print(f"Produção média: {producao_media:.2f} peças por turno.")

print(f"Quantidade de turnos que atingiram a meta: {turno_meta}")

percentual_meta = (turno_meta / valor_turno) * 100
print(f"Percentual de turnos que atingiram a meta: {percentual_meta}%")

```
