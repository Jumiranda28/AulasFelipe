Problema 1: controle de estoque

```python
estoque = [18, 5, 0, 32, 4, 7, 0, 14, 3]

esgotado = 0

critico = 0

for produto in estoque:
    if produto > 5:
        print(f"O produto tem {produto} unidades em estoque. Estoque disponível! ")
    elif produto == 0:
        esgotado += 1
        print(f"O produto não contém unidades em estoque. Estoque esgotado! ")
    else:
        critico += 1
        print(f"O produto tem {produto} unidades em estoque. Estoque crítico! ")

print(f"Contém {esgotado} produtos esgotados!")
print(f"Contém {critico} produtos em estado crítico ")

percentual = ( (esgotado + critico) / len(estoque) )*100

print(f"Percentual de produtos que precisam de reposição:{percentual:.2f} \nConsiderando os em estado crítico e esgotado ")

```
