Problema 1 - Desenpenho de uma rede de supermercados

```python
faturamentos = [18.500, 21.300, 17.800, 24.600, 27.400]
```

```python
quant = len(faturamentos)
print("Quantidade de dias analizados:", quant)
```

```python
total_faturamento = sum(faturamentos)
print(f"Total de faturamento: {total_faturamento:.3f}")
```

```python
media_faturamento = total_faturamento / quant
print(f"Faturamento médio diário: {media_faturamento:.3f}")
```

```python
maior_faturamento = max(faturamentos)
print(f"Maior faturamento: {maior_faturamento:.3f}")
```

```python
menor_faturamento = min(faturamentos)
print(f"Menor faturamento: {menor_faturamento:.3f}")
```

```python
if media_faturamento >= 22.000:
    print("A unidade apresetou um bom desempenho")
elif media_faturamento> 19.000 and media_faturamento < 22.000:
    print("A unidade apresentou um desempenho regular")
else:
    print("A unidade apresentou um desempenho ruim")
```

Problema 3 - Monitoramento de produção industrial

```python
turnos = [485, 510, 472, 498, 455, 520]

quantidade_turnos = len(turnos)
print(f"Quantidade de turnos: {quantidade_turnos}")

total_turnos = sum(turnos)
print(f"Produção total: {total_turnos}")

media_turnos = total_turnos / quantidade_turnos
print(f"Média de produção por turno: {media_turnos:.3f}")

maior_turno = max(turnos)
menor_turno = min(turnos)
print(f"Maior produção em um turno: {maior_turno}, Menor produção em um turno: {menor_turno}")

if media_turnos >= 500:
    print("A meta de produção foi atingida.")
elif media_turnos < 480:
    print("A meta de produção está abaixo do esperado.")
else:
    print("A meta de produção está dentro do esperado, Fiquem atentos para melhorar a produção.")

if menor_turno < 460:
    print("Alerta Operacional!\nO turno está abaixo do esperado, é necessário investigar o motivo da baixa produção.")
```
