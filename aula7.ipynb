# Pratica - registros de participantes

```python
def registrar_treinamentos(tema, *participantes):
    print(f"Tema: {tema}")
    print(f"Quantidade de participantes: {len(participantes)}")
    print("Participantes:")

    for numero, participante in enumerate(participantes, start=1):
        print(f"{numero}. {participante}")

    
```

```python
registrar_treinamentos(
    "Segurança da Informação",
    "Julia",
    "Mohammad",
    "Juan"
)
```

# registros de ocorrencia

```python
def resgistrar_ocorrencia(codigo, *sistemas, **informacoes):
    print("Código:", codigo)
    print(f"Quantidade de sistemas afetados: {len(sistemas)}")
    if len(sistemas) > 0 :
        print("sistemas:")
        for numero, sistema in enumerate(sistemas, start=1):
                print(f"{numero}. {sistema}")
    
    if informacoes:
        print("Informações adicionais:")
        for chave, valor in informacoes.items():
            print(f"{chave}: {valor}")
        
```

```python
resgistrar_ocorrencia("001", "API", prioridade="Alta", responsavel="Mohammad", status="Andamento")
```

## 1. Analise de vendas por vendedor

```python
import statistics
```

```python
def analise_vendedor(nome, *vendas):
    print(f"Nome do vendedor: {nome}")
    valor_total = sum(vendas)
    media_vendas = statistics.mean(vendas)
    maior_valor = max(vendas)
    print("\n===== ANÁLISE DO VENDEDOR ============")
    print(f"Média: R$ {media_vendas:.2f}")
    print(f"Maior valor: R$ {maior_valor:.2f}")
    print(f"Valor Total: R${valor_total:.2f}")
    if valor_total >= 5000.00:
        print(f"A meta do vendedor foi atingida!")
    else:
        print(f"A meta não foi atingida!")
```

```python
analise_vendedor("José", 2000, 300, 150, 450, 500, 1000, 270, 1500 )
```

```python
analise_vendedor("Pedro", 350, 1580, 600, 800, 150 )
```

```python
analise_vendedor("Lucas", 2000, 1590, 450, 800)
```

## 2. Registro de atendimento

```python
def registrar_paciente(Nome, *procedimentos, **informacoes):
     print("Nome do paciente:", Nome)
     print("Quantidade de procedimentos:", len(procedimentos))
     if len(procedimentos) > 0:
          print("\n===== PROCEDIMENTOS REALIZADOS ============")
          for numero, procedimento in enumerate(procedimentos, start=1):
               print(f"{numero}. {procedimento}")
     if informacoes:
          print("\n===== INFORMAÇÕES ADICIONAIS ============")
          for chave, valor in informacoes.items():
               print(f"{chave}: {valor}")  
          if "retorno" in informacoes:
            if informacoes["retorno"]:
               print("\nRetorno: MARCADO")
            else:
               print("\nRetorno: NÃO MARCADO")
          else:
               print("\nRetorno: NÃO INFORMADO") 
            
    
```

```python
registrar_paciente(
    "Maria Silva",
    "consulta",
    "avaliação",
    "radiografia",
    profissional="Dr. Carlos",
    responsavel="Enfermagem",
    unidade_atendimento="Unidade Central",
    convenio="Unimed",
    data_retorno="15/10/2026",
    retorno=True,
    observacao="Paciente apresentou melhora."
)
```

```python
registrar_paciente(
    "João Santos",
    "consulta",
    "exame laboratorial",
    "fisioterapia",
    profissional="Dra. Ana",
    responsavel="Recepção",
    unidade_atendimento="Unidade Norte",
    convenio="Amil",
    retorno=False,
    observacao="Aguardando resultado dos exames."
)
```

## Cadastro de funcionario

```python
def cadastrar_funcionario(nome, idade, salario, setor):
    if not nome:
        return "Nome é obrigatorio"
    if not isinstance(nome, str):
        return "Nome é obrigatorio"
    if 18 <= idade and idade <= 70:
        return "Idade inválida"
    if salario <= 0:
        return "Salario inválido"
    if setor in ["TI", "RH", "Finaceiro", "Comercial"]:
        return "Setor inválido"

    return "Funcionario cadastrado com sucesso"
```
