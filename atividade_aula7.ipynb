### Situação Problema:
Uma empresa utiliza um sistema interno para registrar chamados de suporte. Cada chamado possui
um código, um responsável, uma data de abertura, uma data de fechamento e uma prioridade. Os
dados foram armazenados em uma lista de dicionários.



```python
chamados = [       
    {
        "codigo": "CH001",
        "responsavel": "Maria",
        "abertura": "15/09/2026",
        "fechamento": "17/09/2026",
        "prioridade": "Alta"
    },
    {
        "codigo": "CH002",
        "responsavel": "Carlos",
        "abertura": "16/09/2026",
        "fechamento": "22/09/2026",
        "prioridade": "Média"
    },
    {
        "codigo": "CH003",
        "responsavel": "Marina",
        "abertura": "18/09/2026",
        "fechamento": "19/09/2026",
        "prioridade": "Alta"
    },
    {
        "codigo": "CH004",
        "responsavel": "Rafael",
        "abertura": "20/09/2026",
        "fechamento": "25/09/2026",
        "prioridade": "Baixa"
    },
    {
        "codigo": "CH005",
        "responsavel": "Maria",
        "abertura": "21/09/2026",
        "fechamento": "23/09/2026",
        "prioridade": "Média"
    }
]
```

```python
from datetime import datetime, timedelta

total_chamados = len(chamados)
dentro_prazo = 0
fora_prazo = 0

alta = 0
media = 0
baixa = 0

maior_tempo = 0
codigo_maior_tempo = ""

fora_do_prazo = []

chamados_por_prioridade = {
    "Alta": [],
    "Média": [],
    "Baixa": []
}

for numero, chamado in enumerate(chamados, start=1):

    abertura = datetime.strptime(chamado['abertura'], '%d/%m/%Y')
    fechamento = datetime.strptime(chamado['fechamento'], '%d/%m/%Y')

    tempo = (fechamento - abertura).days

    if tempo <= 3:
        situacao = "Dentro do prazo"
        dentro_prazo += 1
    else:
        situacao = "Fora do prazo"
        fora_prazo += 1
        fora_do_prazo.append({
            "codigo": chamado["codigo"],
            "responsavel": chamado["responsavel"],
            "tempo": tempo
        })

    if chamado['prioridade'] == 'Alta':
        alta += 1
    elif chamado['prioridade'] == 'Média':
        media += 1
    elif chamado['prioridade'] == 'Baixa':
        baixa += 1

    if tempo > maior_tempo:
        maior_tempo = tempo
        codigo_maior_tempo = chamado['codigo']

    chamados_por_prioridade[chamado['prioridade']].append(chamado['codigo'])

    print(f"{numero} | {chamado['codigo']} | {chamado['responsavel']} | {chamado['prioridade']} | {tempo} dias | {situacao}")
    print("=" * 60)
```

```python
print(f"===== RESUMO DOS ATENDIMENTOS ==============================")
print(f"Quantidade total de chamados: {total_chamados}")
print(f"Chamados dentro do prazo: {dentro_prazo}")
print(f"Chamados fora do prazo: {fora_prazo}")
print(f"Código com maior tempo de atendimento: {codigo_maior_tempo}")
print(f"Maior tempo de atendimento: {maior_tempo} dias")
print(f"Prioridade Alta: {alta}")
print(f"Prioridade Média: {media}")
print(f"Prioridade Baixa: {baixa}")
```

```python
print("===== CHAMADOS POR PRIORIDADE ==============================")
for prioridade, codigos in chamados_por_prioridade.items():
    print(prioridade, ":", codigos)
```

```python
print("\n===== CHAMADOS FORA DO PRAZO ===============================")
for chamado in fora_do_prazo:
    print(f"Código: {chamado['codigo']}, Responsável: {chamado['responsavel']}, Tempo: {chamado['tempo']} dias")
```

##  2: Análise de Transações Bancárias


### Situação-problema:

Um banco deseja analisar um conjunto de transações realizadas por seus clientes ao longo de um
dia. Para cada transação são registrados o código, o cliente, o tipo da operação, o valor, o horário
em que a operação foi iniciada e seu tempo de processamento. Os dados foram armazenados em
uma lista de dicionários.

```python
transacoes = [
    {
        "codigo": "TR001",
        "cliente": "Maria",
        "tipo": "PIX",
        "valor": 1250.00,
        "horario": "08:15",
        "processamento": 2
    },
    {
        "codigo": "TR002",
        "cliente": "Carlos",
        "tipo": "Transferência",
        "valor": 3500.00,
        "horario": "09:40",
        "processamento": 8
    },
    {
        "codigo": "TR003",
        "cliente": "Marina",
        "tipo": "Pagamento",
        "valor": 850.00,
        "horario": "10:25",
        "processamento": 5
    },
    {
        "codigo": "TR004",
        "cliente": "Rafael",
        "tipo": "PIX",
        "valor": 1250.00,
        "horario": "11:50",
        "processamento": 3
    },
    {
        "codigo": "TR005",
        "cliente": "Beatriz",
        "tipo": "TED",
        "valor": 4200.00,
        "horario": "13:35",
        "processamento": 15
    },
    {
        "codigo": "TR006",
        "cliente": "Lucas",
        "tipo": "PIX",
        "valor": 2100.00,
        "horario": "14:20",
        "processamento": 2
    },
    {
        "codigo": "TR007",
        "cliente": "Fernanda",
        "tipo": "Pagamento",
        "valor": 850.00,
        "horario": "15:45",
        "processamento": 6
    }
]
```

```python
import statistics

valores = []

for numero, transacao in enumerate(transacoes, start=1):

    horario_inicio = datetime.strptime(
        transacao["horario"], "%H:%M"
    )

    horario_conclusao = horario_inicio + timedelta(
        minutes=transacao["processamento"]
    )
    
    valores.append(transacao["valor"])
    
    print(
        f"{numero}- {transacao['codigo']} | "
        f"{transacao['cliente']} | "
        f"{transacao['tipo']} | "
        f"R$ {transacao['valor']:.2f} | "
        f"{horario_inicio.strftime('%H:%M')} | "
        f"Conclusão: {horario_conclusao.strftime('%H:%M')}"
    )
```

```python
print("\n===== ANÁLISE ESTATÍSTICA ============")

print(f"Média: R$ {statistics.mean(valores):.2f}")
print(f"Mediana: R$ {statistics.median(valores):.2f}")
print(f"Moda: {statistics.multimode(valores)}")
print(f"Variância populacional: {statistics.pvariance(valores):.2f}")
print(f"Desvio padrão populacional: {statistics.pstdev(valores):.2f}")
print(f"Menor valor: R$ {min(valores):.2f}")
print(f"Maior valor: R$ {max(valores):.2f}")
```

```python
media = statistics.mean(valores)
print("\n===== CLASSIFICAÇÃO DAS TRANSAÇÕES ===============")
for numero, transacao in enumerate(transacoes, start=1):

    if transacao["valor"] > media:
        classificacao = "Acima da média"
    elif transacao["valor"] == media:
        classificacao = "Na média"
    else:
        classificacao = "Abaixo da média"
    print(
        f"{numero}- {transacao['codigo']} | "
        f"{transacao['cliente']} | "
        f"R$ {transacao['valor']:.2f} | "
        f"{classificacao}"
    )
```
